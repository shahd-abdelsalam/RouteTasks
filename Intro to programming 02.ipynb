1. Write a function that takes a list of strings and a number K. The function should
return a new list containing only the strings that have a length greater than K.<br>
<br> Input: list=["apple", "pie", "banana"], K=3
<br> Output: ['apple', 'banana']

```python
def text(strings: list[str], k: int):
    """
        Returns a new list containing only the strings whose length is greater than k.
    """
    words = []
    for i in strings:
        if len(i) > k:
            words.append(i)
    return words

l1 = ["apple", "pie", "banana"]
k = 3
print(text(l1, k))

```

2. Write a function named area() that calculates the area of either a "circle" or a
"rectangle". The function should take the shape name as the first argument,
followed by the necessary dimensions (radius for circle; length and width for
rectangle).<br>
<br> Example Input: area("circle", 5) OR area("rectangle", 4, 6)
<br> Example Output: Circle Area: 78.5, Rectangle Area: 24

```python
def area(shape, *dims):
    """
        Calculates the area of a circle or a rectangle based on the given dimensions.
    """
    shape = shape.lower()
    
    if shape == "circle":
        radius = dims[0]
        result = 3.14 * radius **2

        return f"Circle Area: {result}"
    
    elif shape == "rectangle":
        length, width = dims
        result = length * width

        return f"Rectangle Area: {result}"

    else:
        return "shape must be 'circle' or 'rectangle'"


print(area("circle", 5))        
print(area("rectangle", 4, 6))
```

4. Write a program to find the sum of all numbers in a list. ensure
the program ignores any elements in the list that are not numbers (like strings or
booleans).<br>
<br> Example Input: [10,"apple", 20, True, 5]
<br> Example Output: Total Sum: 35

```python
data = [10, "apple", 20, True, 5]

total = 0
for item in data:
    if type(item) == int or type(item) == float:
        total += item

print(f"Total Sum: {total}")
```

5. Write a Python class Restaurant with attributes like menu_items, book_table, and
customer_orders, and methods like add_item_to_menu, book_tables, and
customer_order. (Note: Use dictionaries and lists to store the data.)<br><br>
Perform the following tasks now:<br>
<br> Now add items to the menu.
<br> Make table reservations.
<br> Take customer orders.
<br> Print the menu.
<br> Print table reservations.<br> Print customer orders.

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}       
        self.book_table = {}        
        self.customer_orders = {}   

    def add_item_to_menu(self, name, price):
        self.menu_items[name] = price
        print(f"Added to menu: {name} - ${price}")

    def book_tables(self, table_number, customer_name):
        if table_number in self.book_table:
            print(f"Table {table_number} is already booked.")
        else:
            self.book_table[table_number] = customer_name
            print(f"Table {table_number} booked for {customer_name}.")

    def customer_order(self, table_number, item):
        if table_number not in self.book_table:
            print(f"Table {table_number} is not booked.")
        elif item not in self.menu_items:
            print(f"'{item}' is not on the menu.")
        else:
            self.customer_orders.setdefault(table_number, []).append(item)
            print(f"Table {table_number} ordered: {item}")

    def print_menu(self):
        print("\nMenu:")
        for name, price in self.menu_items.items():
            print(f"{name}: ${price}")
    
    def print_reservations(self):
        print("\nTable Reservations:")
        for table, name in self.book_table.items():
            print(f"Table {table}: {name}")

    def print_orders(self):
        print("\nCustomer Orders:")
        for table, items in self.customer_orders.items():
            print(f"Table {table}: {', '.join(items)}")


r = Restaurant()
r.add_item_to_menu("Pizza", 12)
r.add_item_to_menu("Burger", 9)
r.add_item_to_menu("Pasta", 11)
r.add_item_to_menu("Juice", 4)

r.book_tables(1, "Ahmed")
r.book_tables(2, "Sara")
r.book_tables(1, "Omar")   

r.customer_order(1, "Pizza")
r.customer_order(1, "Juice")
r.customer_order(2, "Burger")
r.customer_order(3, "Pasta")   

r.print_menu()
r.print_reservations()
r.print_orders()
```

<br> requestBike: Takes a request from the customer for the number of bikes and check if
the requested number is available or not if yes, function will deduct number_of_bikes
from shop stock else it will print wait 10 mins. And you have to make sure that the age
of the customer is over than 6 years.
<br> returnBike: Allows customers to return their bikes to the rental shop by taking
rentalTime, num_of_bikes as parameters. And calculate the total bill cost by multiplying

class Bike that is composed of the following attributes:
<br> name
<br> price
Here are some of the methods that you need to add to your class Bike;<br>
<br> def __init__(self).
<br> Getter / setter for all attributes.

```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age

class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes):
        self.bikes = bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, rentals):
        self.rentals = rentals

    def addBike(self, bike):
        self.bikes.append(bike)

    def requestBike(self, customer, num_of_bikes):
        if customer.get_age() <= 6:
            print("Customer must be older than 6 years")
        elif num_of_bikes <= len(self.bikes):
            for i in range(num_of_bikes):
                self.rentals.append(self.bikes.pop())
        else:
            print("Wait 10 mins")

    def returnBike(self, rentalTime, num_of_bikes):
        for i in range(num_of_bikes):
            self.bikes.append(self.rentals.pop())
        return rentalTime * 40

    def totalCost(self):
        total = 0
        for bike in self.bikes:
            total += bike.get_price()
        return total

class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price

shop = MainBikeRental()
shop.addBike(Bike("Mountain", 500))
shop.addBike(Bike("Road", 400))

c = Customer("Ahmed", 25)
shop.requestBike(c, 1)
print(shop.returnBike(3, 1))
print(shop.totalCost())        
```
