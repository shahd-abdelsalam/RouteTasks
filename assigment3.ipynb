1. Create a 3x3 array with random integers between 1 and 100. Write a program to
"normalize" the array so that the values are scaled between 0 and 1. <br>
• Formula: (x - min) / (max - min) <br>
• Example Output: An array where the smallest value becomes 0.0 and the largest becomes 1.0.


````python
import numpy as np

num = np.random.randint(1, 101, (3, 3))
print("Original:\n", num)
if num.max() == num.min():
    normalized = np.zeros_like(num, dtype=float)  # avoid division by zero
else:
    normalized = (num - num.min()) / (num.max() - num.min())
print("Normalized:\n", normalized)
````

2. Create a 1D array of 10 random numbers. Ask the user for a value X Find the value
in the array that is closest to X <br>
• Example Input: Array: [1, 5, 8, 12], X = 10 <br>
• Example Output: 8 or 12 (whichever is mathematically closer).

````python
arr = np.random.randint(1, 101, 10)
print("Array:", arr)

X = float(input("Enter a value X: "))
idx = np.abs(arr - X).argmin()                  # index of smallest distance

print("Closest value to", X, "is", arr[idx])
````

3. Write a NumPy program to create a 5x5 array with random values. For each column
in the array, find the second-largest value. <br>
• Example Input: [[1, 2, 3], [4, 5, 6], [7, 8, 9]] <br>
• Example Output: [4, 5, 6] (These are the second-highest values for each respective column).


````python
b = np.random.randint(1, 100, (5, 5))
print("Array:\n", b)

second_largest = np.sort(b, axis=0)[-2]         # sort each column, take second from the end

print("Second-largest per column:", second_largest)
````

4. Create a 6x4 matrix of random integers. Identify which rows have a sum that is an even number. Print the indices of these rows.

````python
c = np.random.randint(1, 100, (6, 4))
print("Matrix:\n", c)

row_sums = c.sum(axis=1)
print("Row sums:", row_sums)

even_rows = np.where(row_sums % 2 == 0)[0]
print("Indices of rows with even sum:", even_rows)
````

5. Create a 1D array of 30 random integers between 1 and 50. Replace all numbers
that are multiples of 3 or 5 with the value -1. <br>
• Example Input: [3, 7, 10, 15] <br>
• Example Output: [-1, 7, -1, -1]


````python
d = np.random.randint(1, 51, 30)
print("Original:", d)

d[(d % 3 == 0) | (d % 5 == 0)] = -1            
print("Modified:", d) 
````

6. Write a NumPy program to create a 3x3 array with random values. For each row,
calculate its mean (average) and subtract that mean from every element in that
specific row. <br>
• Example Input: ``` [[10, 20, 30], [5, 5, 5], [0, 10, 20]] <br>
• Example Output: [[-10. 0. 10.], [ 0. 0. 0.], [-10. 0. 10.]]

````python
e = np.random.randint(0, 50, (3, 3))
print("Original:\n", e)

result = e - e.mean(axis=1, keepdims=True)    
print("After subtracting row means:\n", result)
 
ex = np.array([[10, 20, 30], [5, 5, 5], [0, 10, 20]])
print("Example check:\n", ex - ex.mean(axis=1, keepdims=True))
````
